# Training YOLO


```python
src = r'/content/drive/My Drive/OREN/Python/Object Detection Project/codes/Training thermal camera dataset/yolo_dataset_output/data.yaml'
des = r'/content/drive/My Drive/OREN/Python/Object Detection Project/codes/Training thermal camera dataset/run2'

# Load the YOLO26 Nano model (best for speed/testing)
model = YOLO('yolo26s.pt')

# Train
results = model.train(
    data=src,
    imgsz=640,
    epochs=200, patience=60, batch=16,
    optimizer="AdamW", lr0=0.001, lrf=0.01, momentum=0.9,
    weight_decay=0.0005, cos_lr=True,
    warmup_epochs=5, warmup_bias_lr=0.01,
    hsv_h=0.0, hsv_s=0.0, hsv_v=0.4,
    degrees=10.0, translate=0.1, scale=0.4, shear=0.0, perspective=0.0,
    fliplr=0.5, flipud=0.5,
    mosaic=1.0, close_mosaic=15, mixup=0.1, copy_paste=0.0,
    cache="ram", seed=0, deterministic=True, plots=True,
    device=0,   # use 0 on Colab
)
```

```python
import cv2 as cv
import numpy as np
import os
from google.colab import drive
from pathlib import Path
from numpy.random import randint

!pip install ultralytics
from ultralytics import YOLO

drive.mount('/content/drive', force_remount=True)
```

```python
if os.path.exists('/content/drive/MyDrive'):
    print("Success: Drive is mounted and files are accessible.")
else:
    drive.mount('/content/drive', force_remount=True)
    print("Failure: Drive is not mounted.")
```

```python
src1 = r'/content/drive/My Drive/OREN/Python/Object Detection Project/codes/Training thermal camera dataset/run1/weights/best.pt'
src2 = r'/content/drive/My Drive/OREN/Python/Object Detection Project/codes/Training thermal camera dataset/run2/detect/train/weights/best.pt'
data_yaml_path = r'/content/drive/My Drive/OREN/Python/Object Detection Project/codes/Training thermal camera dataset/yolo_dataset_output/data.yaml'
des1 = r'/content/drive/My Drive/OREN/Python/Object Detection Project/codes/Training thermal camera dataset/results/yolo26n_run_3_0.5'
des2 = r'/content/drive/My Drive/OREN/Python/Object Detection Project/codes/Training thermal camera dataset/results/yolo26s_run_3_0.5'

path_to_check = Path(data_yaml_path)

if path_to_check.exists():
    print("Success: The path exists!")
else:
    print("Error: The path does not exist. Check your spelling or Drive mount.")
```

```python
model1 = YOLO(src1)
model2 = YOLO(src2)


results1 = model1.val(data=data_yaml_path,split="test", batch=16, conf=0.7, iou=0.3, device="0",project=des1,plots=True)
results2 = model2.val(data=data_yaml_path,split="test", batch=16, conf=0.3, iou=0.7, device="0",project=des2,plots=True)

print('***** MODEL 1 ******\n')
print(results1.confusion_matrix.to_df())

print('\n***** MODEL 2 ******\n')
print(results2.confusion_matrix.to_df())
```
